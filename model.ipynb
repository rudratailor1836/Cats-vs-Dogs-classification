```python
import tensorflow as tf
from tensorflow import keras 
from keras.models import Sequential 
from keras.layers import Dense,Conv2D,MaxPooling2D,Flatten 
```

```python
# generrators to give data to model in batch as data is too big to fit in the RAM
train_ds = keras.utils.image_dataset_from_directory(
    directory='/workspaces/Cats-vs-Dogs-classification/train',
    labels='inferred',
    label_mode='int',
    batch_size=128,
    image_size=(256,256)
)

```

```python
validation_ds = keras.utils.image_dataset_from_directory(directory='/workspaces/Cats-vs-Dogs-classification/validation_data',labels='inferred',label_mode='int',batch_size=128,image_size=(256,256))
```

```python
#  Normalize the data

def process(img,label):
    img = tf.cast(img/255. ,tf.float32 )
    return img,label

train_ds = train_ds.map(process)
validation_ds = validation_ds.map(process)
```

```python
# create cnn model 
model = Sequential()

model.add(Conv2D(32,kernel_size=(3,3),padding='valid',activation='relu',input_shape=(256,256,3)))
model.add(MaxPooling2D(pool_size=(2,2),strides=2,padding='valid'))

model.add(Conv2D(64,kernel_size=(3,3),padding='valid',activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2),strides=2,padding='valid'))

model.add(Conv2D(128,kernel_size=(3,3),padding='valid',activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2),strides=2,padding='valid'))

model.add(Flatten())

model.add(Dense(128, activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
hist = model.fit(train_ds,epochs=10,validation_data='validation_ds')
```
